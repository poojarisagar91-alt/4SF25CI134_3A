```python
import pandas as pd
import zipfile

with zipfile.ZipFile("archive (1).zip", "r") as z:
    csv_file = next(
        name for name in z.namelist()
        if name.lower().endswith(".csv")
    )

    with z.open(csv_file) as f:
        df = pd.read_csv(f)
```

```python


# Count all missing values
print("Total null values:", df.isnull().sum().sum())
```

```python
numeric_columns = df.select_dtypes(include="number").columns

print("Numerical columns:")
print(numeric_columns.tolist())
```

```python
for col in numeric_columns:
    if df[col].isnull().any():
        if df[col].notna().any():
            df[col] = df[col].fillna(df[col].mean())

print("Null values after mean imputation:")
print(df.isnull().sum())
```

```python
print("Total null values remaining:", df.isnull().sum().sum())
```

```python
print(df.columns.tolist())
```

```python
df = df.loc[:, ~df.columns.str.startswith("Unnamed:")]

print("Updated shape:", df.shape)
print(df.head())
```

```python
print(df.dtypes)
```

```python
if "popularity" in df.columns:
    df["popularity"] = pd.to_numeric(
        df["popularity"], errors="coerce"
    )

print(df.dtypes)
```

```python
if "duration_ms" in df.columns:
    df["duration_minutes"] = df["duration_ms"] / 60000

print(df.head())
```

```python
print(df.columns.tolist())
print(df.describe(include="all").T)
```

```python
print(df.select_dtypes(include="number").columns.tolist())
```

```python
print("Mean values:")
print(df.select_dtypes(include="number").mean())

print("\nMinimum values:")
print(df.select_dtypes(include="number").min())

print("\nMaximum values:")
print(df.select_dtypes(include="number").max())
```

```python
for col in df.select_dtypes(include="object").columns[:5]:
    print("\nMost common values in", col)
    print(df[col].value_counts().head(5))
```

```python
import matplotlib.pyplot as plt

numeric_columns = df.select_dtypes(include="number").columns

if len(numeric_columns) > 0:
    df[numeric_columns[0]].dropna().hist(bins=20)

    plt.title("Distribution of " + numeric_columns[0])
    plt.xlabel(numeric_columns[0])
    plt.ylabel("Frequency")
    plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")

if numeric_df.shape[1] >= 2:
    correlation = numeric_df.corr()

    plt.figure(figsize=(10, 6))
    plt.imshow(correlation, cmap="coolwarm", aspect="auto")
    plt.colorbar(label="Correlation")
    plt.xticks(range(len(correlation.columns)),
               correlation.columns, rotation=90)
    plt.yticks(range(len(correlation.columns)),
               correlation.columns)
    plt.title("Correlation Between Numerical Columns")
    plt.tight_layout()
    plt.show()
```
