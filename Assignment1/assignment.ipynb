```python

# Q1: Count the number of words in a file

def count_words():
    # Open the file in read mode
    file_path = r"C:\Users\SAGAR\Downloads\skill60 assignment2\document.txt"

    with open(file_path, "r", encoding="utf-8") as file:
        text = file.read()

    # Split the text into words and count them
    return len(text.split())


# Display the result
print("Number of words:", count_words())
```

```python

# Q2: Find and replace text in a file

def replace_word(old_word, new_word):
    file_path = r"C:\Users\SAGAR\Downloads\skill60 assignment2\data.txt"

    # Read the file
    with open(file_path, "r", encoding="utf-8") as file:
        text = file.read()

    # Replace the old word with the new word
    text = text.replace(old_word, new_word)

    # Write the updated text back to the file
    with open(file_path, "w", encoding="utf-8") as file:
        file.write(text)

    print("Text replaced successfully.")


# Replace Python with Java
replace_word("Python", "Java")
```

```python

# Q3: Read a file in reverse order

def read_reverse():
    file_path = r"C:\Users\SAGAR\Downloads\skill60 assignment2\reverse.txt"

    # Open the file in read mode
    with open(file_path, "r", encoding="utf-8") as file:
        lines = file.readlines()

    # Display lines in reverse order
    for line in reversed(lines):
        print(line, end="")


# Call the function
read_reverse()
```

```python

# Q4: Calculate file statistics

def file_statistics():
    file_path = r"C:\Users\SAGAR\Downloads\skill60 assignment2\stats.txt"

    # Read the file
    with open(file_path, "r", encoding="utf-8") as file:
        text = file.read()

    # Calculate statistics
    lines = len(text.splitlines())
    words = len(text.split())
    characters = len(text)

    # Display the results
    print("Number of lines:", lines)
    print("Number of words:", words)
    print("Number of characters:", characters)


# Call the function
file_statistics()
```

```python

# Q6: Replace diagonal elements with 0

import numpy as np

# Create an array containing numbers from 1 to 16
arr = np.arange(1, 17).reshape(4, 4)

# Display the original array
print("Original array:")
print(arr)

# Replace diagonal elements with 0
np.fill_diagonal(arr, 0)

# Display the modified array
print("Array after replacing diagonal elements:")
print(arr)
```

```python

# Q7: Find common elements between two arrays

import numpy as np

# Create two arrays
arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.array([30, 40, 50, 60, 70])

# Find common elements
common = np.intersect1d(arr1, arr2)

# Display the result
print("First array:", arr1)
print("Second array:", arr2)
print("Common elements:", common)
```

```python

# Q9: Reshape a NumPy array

import numpy as np

# Create a 3x3 array with values from 1 to 9
arr = np.arange(1, 10).reshape(3, 3)

print("Original 3x3 array:")
print(arr)

# Reshape into 1 row and 9 columns
arr1 = arr.reshape(1, 9)
print("\nReshaped array (1x9):")
print(arr1)

# Reshape into 9 rows and 1 column
arr2 = arr.reshape(9, 1)
print("\nReshaped array (9x1):")
print(arr2)
```

```python

# Q10: Create and sort a structured NumPy array by age

import numpy as np

# Define the data types for each field
data_type = [
    ("name", "U10"),
    ("age", "i4"),
    ("marks", "f4")
]

# Create the structured array
students = np.array([
    ("Sagar", 20, 85.5),
    ("Rahul", 19, 90.0),
    ("Anu", 21, 78.5),
    ("Priya", 18, 92.0)
], dtype=data_type)

# Display the original array
print("Original array:")
print(students)

# Sort the array by age
sorted_students = np.sort(students, order="age")

# Display the sorted array
print("\nArray sorted by age:")
print(sorted_students)
```

```python

# Q21: Calculate the sum of each row using lambda

import pandas as pd

# Create a DataFrame
df = pd.DataFrame({
    "A": [10, 20, 30, 40, 50],
    "B": [5, 15, 25, 35, 45],
    "C": [2, 4, 6, 8, 10]
})

# Calculate the sum of each row using lambda
df["Total"] = df.apply(lambda row: row.sum(), axis=1)

# Display the DataFrame
print(df)
```

```python

# Q23: Count vowels in each string

import pandas as pd

# Create a Pandas Series
s = pd.Series(["python", "education", "computer", "assignment"])

# Function to count vowels
def count_vowels(text):
    count = 0

    for char in text.lower():
        if char in "aeiou":
            count += 1

    return count

# Count vowels in each string
vowel_counts = s.apply(count_vowels)

# Display the results
print("Original Series:")
print(s)

print("\nVowel counts:")
print(vowel_counts)
```
